```python
# 1. Class Name: BankAccount
# Problem
# Create a class named BankAccount.
# Explanation
# A bank account stores money. The balance should not be directly accessible from outside
# the class, so create a private attribute named __balance.
# Implement the following methods:
# • deposit(amount)
# o Deposit money into the account.
# o Amount must be greater than 0.
# • withdraw(amount)
# o Withdraw money.
# o Withdrawal should happen only if:
# ▪ amount > 0
# ▪ amount <= available balance

class BankAccount:
    def __init__(self):
        self.__balance = 0   # Private attribute

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Deposit amount must be greater than 0.")

    # Withdraw method
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be greater than 0.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    # Display current balance
    def display_balance(self):
        print(f"Current Balance: ₹{self.__balance}")


# Creating object
account = BankAccount()

# Deposit money
account.deposit(5000)
account.display_balance()

# Withdraw money
account.withdraw(2000)
account.display_balance()

# Invalid cases
account.deposit(-100)
account.withdraw(6000)
account.withdraw(-500)
```

```python
# 2. Class Name: Student
# Problem
# Create a class named Student.
# Explanation
# Every student has marks that should be private.
# Create:
# • Private attribute __marks
# Methods:
# • set_marks(marks)
# o Accept marks only between 0 and 100.
# • get_marks()
#o Return the student's marks.

class Student:
    def __init__(self):
        self.__marks=0
    

    def set_marks(self, marks):
        if marks >=0 and marks<=100:
            self.__marks = marks
            print("Marks.")
        else:
            print("Invalid Marks.")
    def get_marks(self):
        return self.__marks
     
stu1=Student()
stu1.set_marks(10)
print("Student Marks:",stu1.get_marks())
stu1.set_marks(-120)
print("Student Marks:",stu1.get_marks())
```

```python
# 3. Class Name: Employee
# Problem
# Create a class named Employee.
# Explanation
# Salary should not be modified directly.
# Create:
# • Private attribute __salary
# Methods:
# • set_salary(new_salary)
# o Salary cannot decrease by more than 10% of the current salary.
# • get_salary()
# Example:
# Current Salary = 50,000
# Allowed minimum = 45,000
# New salary = 40,000
# Reject

class Employee:
    def __init__(self, salary):
        self.__salary = salary 

    def set_salary(self, new_salary):
        allowed_minimum = self.__salary * 10.0 

        if new_salary >= allowed_minimum:
            self.__salary = new_salary
            print("Salary updated successfully.")
        else:
            print("Salary update rejected! Salary cannot decrease by more than 10%.")

    def get_salary(self):
        return self.__salary

emp = Employee(50000)
print("Current Salary:", emp.get_salary())

emp.set_salary(45000)
print("Updated Salary:", emp.get_salary())

emp.set_salary(40000)
print("Current Salary:", emp.get_salary())
```

```python
# 4. Class Name: Mobile
# Problem
# Create a class named Mobile.
# Explanation
# Price should always be positive.
# Private attribute:
# • __price
# Methods:
# • set_price(price)

# Sumit Sir Notes
# • get_price()
# Validation:
# Price > 0

class Mobile:
    def __init__(self):
        self.__price = 0 

    
    def set_price(self, price):
        if price > 0:
            self.__price = price
            print("Price updated successfully.")
        else:
            print("Invalid price! Price must be greater than 0.")

    def get_price(self):
        return self.__price

mobile = Mobile()

mobile.set_price(25000)
print("Mobile Price:", mobile.get_price())

mobile.set_price(-5000)
print("Mobile Price:", mobile.get_price())
```

```python
# 7. Class Name: Car
# Problem
# Create class Car.
# Explanation
# Store privately
# • __model
# • __year
# Provide
# • get_model()
# • set_model()
# • get_year()
# • set_year()
# Validation:
# Year cannot be greater than current year.

from datetime import datetime

class Car:
    def __init__(self, model, year):
        self.__model = model      # Private attribute
        self.__year = year        # Private attribute

    # Getter for model
    def get_model(self):
        return self.__model

    # Setter for model
    def set_model(self, model):
        self.__model = model
        print("Model updated successfully.")

    # Getter for year
    def get_year(self):
        return self.__year

    # Setter for year
    def set_year(self, year):
        current_year = datetime.now().year
        if year <= current_year:
            self.__year = year
            print("Year updated successfully.")
        else:
            print("Invalid year! Year cannot be greater than the current year.")


# Creating object
car = Car("Toyota", 2022)

# Display details
print("Model:", car.get_model())
print("Year:", car.get_year())

# Update model
car.set_model("Honda")

# Update year (Valid)
car.set_year(2025)

# Update year (Invalid)
car.set_year(2030)

# Display updated details
print("Updated Model:", car.get_model())
print("Updated Year:", car.get_year())
```
