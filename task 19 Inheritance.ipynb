```python
class BankAccount:
    def __init__(self, account_holder: str, balance: float = 0.0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount: float):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount: float):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}")
        elif amount > self.balance:
            print("Insufficient funds!")
        else:
            print("Withdrawal amount must be positive.")


class SavingsAccount(BankAccount):
    def __init__(self, account_holder: str, balance: float, interest_rate: float):
        # Initialize the parent class attributes
        super().__init__(account_holder, balance)
        self.interest_rate = interest_rate  # Represented as a decimal (e.g., 0.05 for 5%)

    def calculate_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Interest added: ${interest:.2f}. New balance: ${self.balance:.2f}")


# --- Demonstration Code ---
if __name__ == "__main__":
    print("--- Testing SavingsAccount ---")
    # Create a savings account with a $1000 balance and 5% interest rate
    savings = SavingsAccount("Alice", 1000.0, 0.05)
    
    savings.deposit(500)       # Deposited $500.00. New balance: $1500.00
    savings.withdraw(200)      # Withdrew $200.00. New balance: $1300.00
    savings.calculate_interest()
```

```python
# 12. Class Name: Account → LoanAccount
# Problem
# Create a parent class named Account.
# Explanation
# Store:
# • balance
# Create subclass:
# • LoanAccount
# Add:
# • calculate_emi()
# class Account:
#     def __init__(self,balance):
#         self.balance=balance
# class LoanAccount(Account):
#     def calculate_emi(self):

class Account:
    def __init__(self, balance):
        self.balance=balance


class LoanAccount(Account):
    def __init__(self, balance, loan_amount, months):
        super().__init__(balance)
        self.loan_amount=loan_amount
        self.months=months

    def calculate_emi(self):
        emi = self.loan_amount / self.months
        print(f"Account Balance: ₹{self.balance}")
        print(f"Loan Amount: ₹{self.loan_amount}")
        print(f"Loan Duration: {self.months} months")
        print(f"Monthly EMI: ₹{emi:.2f}")


loan=LoanAccount(50000, 120000, 24)
loan.calculate_emi()
```

```python
# 19. Class Name:
# Shape → Polygon → Rectangle
# Problem
# Create multi-level inheritance.
# Explanation
# Rectangle should calculate its area using length and width.

class Shape:
    def __init__(self,name):
        self.name=name

class Polygon(Shape):
    def __init__(self, name, sides):
        super().__init__(name)
        self.sides=sides

class Rectangle(Polygon):
    def __init__(self, length, width):
        super().__init__("Rectangle", 4)
        self.length=length
        self.width=width

    def area(self):
        return self.length * self.width

rect=Rectangle(5,10)

print(f"Shape: {rect.name}")
print(f"Sides: {rect.sides}")
print(f"Area: {rect.area()}")
```

```python
# 20. Class Name:
# BankAccount → SavingsAccount → RecurringDeposit
# Problem
# Create three levels of inheritance.
# Explanation
# Each class should add new functionality.
# Example:
# • deposit
# • interest
# • recurring deposit maturity
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"₹{amount} deposited successfully.")


class SavingsAccount(BankAccount):
    def __init__(self, account_holder, balance, interest_rate):
        super().__init__(account_holder, balance)
        self.interest_rate=interest_rate

    def calculate_interest(self):
        interest=(self.balance * self.interest_rate) / 100
        return interest


class RecurringDeposit(SavingsAccount):
    def __init__(self, account_holder, balance, interest_rate, monthly_deposit, months):
        super().__init__(account_holder, balance, interest_rate)
        self.monthly_deposit=monthly_deposit
        self.months=months

    def maturity_amount(self):
        total_deposit=self.balance + (self.monthly_deposit * self.months)
        interest=(total_deposit * self.interest_rate * self.months) / (12 * 100)
        return total_deposit + interest

rd=RecurringDeposit("Piyush", 10000, 6, 2000, 12)


rd.deposit(5000)

print("\nAccount Holder:", rd.account_holder)
print("Balance:", rd.balance)
print("Interest on Current Balance:", rd.calculate_interest())
print("Recurring Deposit Maturity Amount:", rd.maturity_amount())
```
