```python
#1. Print Numbers from 1 to 20
#Question
#Write a program to print numbers from 1 to 20 using a while loop.
#Explanation
#• Start from 1
#• Print the number
#• Increase the value by 1 each time
#• Stop when number becomes greater than 20

# Print numbers from 1 to 20 using a while loop

num = 1

while num <= 20:
    print(num)
    num += 1
```

```python
#2. Print Even Numbers
#Question
#Write a program to print all even numbers between 1 and 50 using a while
#loop.
#Explanation
#• Even numbers are divisible by 2
#• Use loop from 1 to 50
#• Check condition using if num % 2 == 0
# Print numbers from 1 to 20 using a while loop

num = 1
while num <= 50:
    if num % 2==0:
        print(num)
    num +=1
```

```python
#3. Sum of Natural Numbers
#Question
#Write a program to calculate the sum of first n natural numbers.
#Explanation

#Example:
#1 + 2 + 3 + 4 + 5 = 15
#• Take value of n
#• Use loop from 1 to n
#• Add each number into a sum variable

n = int(input("Enter the value of n: "))

i = 1
sum = 0

while i <= n:
    sum = sum + i
    i = i + 1

print("Sum of first", n, "natural numbers is:", sum)
    
```

```python
#4. Reverse a Number
#Question
#Write a program to reverse a number using a while loop.
#Explanation
#Example:
#1234 → 4321
#Steps:
#• Take last digit using % 10
#• Add digit into reverse number
#• Remove last digit using // 10

num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number is:", reverse)
```

```python
#5. Count Digits in a Number
#Question
#Write a program to count total digits in a number.
#Explanation
#Example:
#56789 → 5 digits
#Steps:
#• Use loop until number becomes 0
#• Remove one digit each time
#• Increase counter by 1

num = int(input("Enter a number: "))

count = 0

while num > 0:
    num = num // 10
    count = count + 1

print("Total digits:", count)
```

```python
#6. Palindrome Number
#Question
#Write a program to check whether a number is palindrome or not.
#Explanation
#Palindrome means same from both sides.
#Example:
#121 → Palindrome
#1331 → Palindrome
#Steps:
#• Reverse the number
#• Compare with original number
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
#7. Armstrong Number
#Question
#Write a program to check whether a number is Armstrong number.
#Explanation
#Example:
#153
#Calculation:
#13 + 53 + 33 = 153
#If sum equals original number → Armstrong number
num = int(input("Enter a number: "))

original = num
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + (digit ** 3)
    num = num // 10

if sum == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
#9. Guessing Game
#Question
#Write a program for number guessing game.
#Explanation
#• Store one secret number
#• Ask user to guess
#• If wrong → ask again
#• If correct → print success message
secret_number = 7

guess = int(input("Guess the number: "))

while guess != secret_number:
    print("Wrong guess! Try again.")
    guess = int(input("Guess the number: "))

print("Congratulations! You guessed the correct number.")
```

```python
#11. Fibonacci Series
#Question
#Write a program to print Fibonacci series.
#Explanation
#Series:
#0 1 1 2 3 5 8 13 ...
#Each next number is sum of previous two numbers.
# Fibonacci Series

n = int(input("Enter the number of terms: "))

a = 0
b = 1
count = 0

print("Fibonacci Series:")

while count < n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    count += 1
```

```python
#12. Factorial of Number
#Question
#Write a program to find factorial of a number.
#Explanation
#Example:
#5! = 5 × 4 × 3 × 2 × 1 = 120
#• Multiply numbers one by one using loop
# Factorial of a Number

num = int(input("Enter a number: "))

fact = 1
i = 1

while i <= num:
    fact = fact * i
    i += 1

print("Factorial =", fact)
```

```python
#13. Menu Driven Calculator
#Question
#Create calculator using while loop and if elif else.
#Explanation
#Menu Example:
#1. Add
#2. Subtract
#3. Multiply
#4. Divide
#5. Exit
#• User selects option
#• Perform operation
#• Continue until exit option chosen
# Menu Driven Calculator

while True:
    print("\n----- Calculator Menu -----")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = int(input("Enter your choice (1-5): "))

    if choice == 5:
        print("Thank you! Calculator closed.")
        break

    if choice >= 1 and choice <= 4:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        if choice == 1:
            print("Result =", num1 + num2)

        elif choice == 2:
            print("Result =", num1 - num2)

        elif choice == 3:
            print("Result =", num1 * num2)

        elif choice == 4:
            if num2 != 0:
                print("Result =", num1 / num2)
            else:
                print("Division by zero is not allowed.")

    else:
        print("Invalid choice! Please enter a number between 1 and 5.")
```

```python
#15. Sum Until User Enters 0
#Question
#Take numbers from user continuously and stop when user enters 0.
#Explanation
#• Add all numbers into sum
#• When input becomes 0 → stop loop3
#• Print final sum
# Sum Until User Enters 0

total = 0

while True:
    num = int(input("Enter a number (0 to stop): "))

    if num == 0:
        break

    total = total + num

print("Final Sum =", total)
```

```python
#16. Leap Year Checker Repeatedly
#Question
#Check leap year repeatedly until user types "stop".
#Explanation
#• Ask year from user
#• Check leap year condition
#• Continue until "stop" entered
#Leap year condition:
#• Divisible by 4
#• Not divisible by 100 OR divisible by 400
# Leap Year Checker Repeatedly

while True:
    year = input("Enter a year (or type 'stop' to exit): ")

    if year.lower() == "stop":
        print("Program stopped.")
        break

    year = int(year)

    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(year, "is a Leap Year.")
    else:
        print(year, "is NOT a Leap Year.")
```

```python
#18. Print multiplication table from 1 to 5 using nested loop.
#Explanation: Outer loop for rows, inner for columns.

#Input: None
#Output: Full 5x5 table
# Print Multiplication Table from 1 to 5 using nested while loops

i = 1

while i <= 5:
    j = 1
    while j <= 5:
        print(i * j, end="\t")
        j += 1
    print()
    i += 1
```

```python
#19. Print prime numbers between 1 to 100 using nested for loop.
#Explanation: Outer loop for numbers, inner loop to check divisibility.
#nput: None
#Output: List of primes
# Print Prime Numbers from 1 to 100 using nested for loops

for num in range(2, 101):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, end=" ")
```
