#### 21. Create a custom exception InsufficientBalanceError for a banking program.

```python
class InsufficientBalanceError(Exception):
    def __init__(self, message="Balance is too low for this action."):
        super().__init__(message)

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Error: Withdrawal amount exceeds current balance!")
        self.balance -= amount
        print(f"Withdrawal successful. Remaining balance: {self.balance}")

account = BankAccount(1000)
try:
    account.withdraw(1500)
except InsufficientBalanceError as e:
    print(e)
```

#### 22. Write a program that raises TypeError manually if the input is not a string.

```python
def validate_string(user_input):
    """Validates if the provided input is a string."""

    if not isinstance(user_input, str):

        raise TypeError(f"Invalid input type: Expected 'str', but got '{type(user_input).__name__}'")
    
    print(f"Success! '{user_input}' is a valid string.")

try:
    validate_string("Hello, World!")
except TypeError as error:
    print(f"Caught an error: {error}")

try:
    validate_string(42)
except TypeError as error:
    print(f"Caught an error: {error}")

try:
    validate_string(["apple", "banana"])
except TypeError as error:
    print(f"Caught an error: {error}")
```

#### 23. Create a function that checks password length and raises a custom exception if length is less than 8.

```python
class PasswordTooShortError(Exception):
    def __init__(self, message="Password must be at least 8 characters long."):
        super().__init__(message)

def validate_password(password):
    if len(password) < 8:
        raise PasswordTooShortError()
    return "Password is valid!"

passwords_to_test = ["abc12", "securePass123"]

for pwd in passwords_to_test:
    try:
        print(f"Testing: {pwd}")
        result = validate_password(pwd)
        print(result)
    except PasswordTooShortError as e:
        print(e)
```

#### 25. Write a program that uses raise to force an exception and handles it.

```python
def trigger_forced_exception():
    try:
        print("1. Entering the try block...")

        raise RuntimeError("This exception was forced on purpose!")
        print("This line will never execute.")
        
    except RuntimeError as error:
        print("2. Entering the except block...")
        print(f"   Handled forced exception: {error}")
        
    finally:
        print("3. Entering the finally block...")
        print("   Execution completed successfully.")

trigger_forced_exception()
```
