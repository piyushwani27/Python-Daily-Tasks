```python
# 1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
# from both B and C).
# Concept
# Hybrid Inheritance (Combination of Single + Multiple Inheritance)
# What to do
# • Create parent class A.
# • Create class B inheriting from A.
# • Create another class C.
# • Create class D inheriting from both B and C.
# How to solve
# 1. Create class A.
# 2. Add one method.
# 3. Create class B(A).
# 4. Create class C.
# 5. Create class D(B, C).
# 6. Create an object of D.
# 7. Call methods from all parent classes.
# Parent class A
class A:
    def display_a(self):
        print("This is Class A")

# Class B inherits from A (Single Inheritance)
class B(A):
    def display_b(self):
        print("This is Class B")

# Another Parent Class C
class C:
    def display_c(self):
        print("This is Class C")

# Class D inherits from both B and C (Multiple Inheritance)
class D(B, C):
    def display_d(self):
        print("This is Class D")

# Creating object of D
obj = D()

# Calling methods from all parent classes
obj.display_a()
obj.display_b()
obj.display_c()
obj.display_d()
```

```python
# 2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
# Concept
# Real-world Hybrid Inheritance
# What to do
# • Person is the base class.
# • Employee inherits Person.
# • Student inherits Person.
# • TeachingAssistant inherits Employee and Student.
# How to solve

# Sumit Sir Notes
# Hierarchy
# Person
# / \
# Employee Student
# \ /
# TeachingAssistant
# Steps
# 1. Store name in Person.
# 2. Employee stores salary.
# 3. Student stores marks.
# 4. TeachingAssistant stores department.
# 5. Display all information.
class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, name, salary):
        Person.__init__(self, name)
        self.salary = salary


class Student(Person):
    def __init__(self, name, marks):
        Person.__init__(self, name)
        self.marks = marks


class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        Employee.__init__(self, name, salary)
        Student.__init__(self, name, marks)
        self.department = department

    def display(self):
        print(f"Name       : {self.name}")
        print(f"Salary     : {self.salary}")
        print(f"Marks      : {self.marks}")
        print(f"Department : {self.department}")


# Creating object
ta = TeachingAssistant("Piyush", 50000, 92, "Computer Science")

# Displaying information
ta.display()
```
