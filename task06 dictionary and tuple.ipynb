### 21. (Hard) Create a dictionary student with 4 key-value pairs. Then use update() to add another dictionary, remove the last item using popitem(), and finally print all items using items().

```python
# Step 1: Create a dictionary
student = {
    "Rahul": 85,
    "Priya": 92,
    "Aman": 78,
    "Neha": 88
}

# Step 2: Add another dictionary using update()
new_student = {
    "Riya": 95
}

student.update(new_student)

# Step 3: Remove the last inserted item
removed_item = student.popitem()
print("Removed Item:", removed_item)

# Step 4: Print all remaining items
print("\nRemaining Student Dictionary:")
for key, value in student.items():
    print(key, ":", value)
```

```python
# dictionary named as the student in a key: value pairs.
student ={
    "Piyush":98,
    "Sahil":89,
    "Sameer":86,
    "Sanket":56
}
print(student) # print the dictionary students. 
# add another dictionary using update().
new_student={
    "Yash":65
}
print(student.update(new_student))
# remove the last inserted item using popitem().
removed_item= student.popitem()
print("Removed item:",removed_item)
# for printing the key value pairs of the student dictionary.
# using for loop.
for key,value in student.items():
    print(key,":",value)
```

### 22. (Hard) Create a tuple with 8 numbers. Find the index of the number 25. If it exists, count how many times it appears. Then convert the tuple into a list, sort it in descending order, and print.

```python

numbers = (10, 25, 30, 40, 25, 50, 60, 70)

# Check if 25 exists
if 25 in numbers:
    print("Index of 25:", numbers.index(25))
    print("Count of 25:", numbers.count(25))
else:
    print("25 is not present in the tuple.")

# Convert tuple to list
num_list = list(numbers)

# Sort the list in descending order
num_list.sort(reverse=True)

# Print the sorted list
print("Sorted List (Descending):", num_list)
```
