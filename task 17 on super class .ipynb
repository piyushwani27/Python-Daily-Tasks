```python
# 1. Class Name: Shape
# Problem
# Create an abstract class named Shape.
# Explanation
# A shape can have different formulas for area and perimeter. Therefore, create two abstract
# methods:
# • area()
# • perimeter()
# Create subclasses like Circle and Rectangle and implement both methods.

from abc import ABC,abstractmethod
class Shape:
    def __init__(self,area,perimeter):
        self.__area=area
        self.__perimeter=perimeter
    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass
        
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def calculate_area(self):
        return 3.14*self.radius** 2

    def calculate_perimeter(self):
        return 3.14*self.radius** 2
        
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length=length
        self.width=width

    def calculate_area(self):
        return self.length*self.width

    def calculate_perimeter(self):
        return 2*(self.length + self.width)
circle = Circle(7)
print("Circle Area:", circle.calculate_area())
print("Circle Perimeter:", circle.calculate_perimeter())

rectangle = Rectangle(10, 5)
print("Rectangle Area:", rectangle.calculate_area())
print("Rectangle Perimeter:", rectangle.calculate_perimeter())
```

```python
from abc import ABC, abstractmethod
import math

# Parent Abstract Class
class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass


# Child Class: Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return math.pi * self.radius ** 2

    def calculate_perimeter(self):
        return 2 * math.pi * self.radius


# Child Class: Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)


# Create Objects
circle = Circle(7)
print("Circle Area:", round(circle.calculate_area(), 2))
print("Circle Perimeter:", round(circle.calculate_perimeter(), 2))

rectangle = Rectangle(10, 5)
print("Rectangle Area:", rectangle.calculate_area())
print("Rectangle Perimeter:", rectangle.calculate_perimeter())
```

```python
# 3. Class Name: Animal
# Problem
# Create an abstract class named Animal.
# Explanation
# Every animal makes a sound, but the sound depends on the animal.
# Create an abstract method:
# • make_sound()
# Create subclasses:
# • Dog
# • Cat
# Each should implement its own sound.
# from abc import ABC, abstractmethod


from abc import ABC,abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

class Dog(Animal):
    def make_sound(self):
        print("Bhauuuuuuu Bhauuuuu uuuuuu")

class Cat(Animal):
    def make_sound(self):
        print("Meaoooooooo Meaooooo Meawoooo.")

my_dog=Dog()
my_dog.make_sound()

my_cat=Cat()
my_cat.make_sound() 
```

```python
# 4. Class Name: Payment
# Problem
# Create an abstract class named Payment.
# Explanation
# Every payment processes money differently.
# Create an abstract method:
# • process_payment(amount)
# Create subclasses:
# • CreditCard
# • Cash
# Each should process the payment in its own way.

from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def process_payment(self,amount):
        pass

class CreditCard(Payment):
    def process_payment(self,amount):
        print(f"Credit Card Payment Processed: {amount}")

class Cash(Payment):
    def process_payment(self, amount):
        print(f"Cash Payment Processed: {amount}")

credit=CreditCard()
cash=Cash()

credit.process_payment(5000)
cash.process_payment(1200)
```

```python
# 5. Class Name: Report
# Problem
# Create an abstract class named Report.
# Explanation
# Different report formats generate reports differently.
# Create an abstract method:
# • generate()
# Create:
# • PDFReport
# • ExcelReport
# Each should generate the report differently.
from abc import ABC, abstractmethod

class Report(ABC):
    @abstractmethod
    def generate(self):
        pass


class PDFReport(Report):
    def generate(self):
        print("PDF Report :")


class ExcelReport(Report):
    def generate(self):
        print("Excel Report :")

pdf=PDFReport()
pdf.generate()

excel=ExcelReport()
excel.generate()
```

```python
# 6. Class Name: BankAccount
# Problem
# Create an abstract class named BankAccount.
# Explanation
# Different bank accounts calculate interest differently.
# Create an abstract method:
# • calculate_interest()
# Create:
# • SavingsAccount
# Implement the interest calculation.

from abc import ABC, abstractmethod
class BankAccount(ABC):
    @abstractmethod
    def calculate_interest(self):
        pass

class SavingsAccount(BankAccount):
    def __init__(self, balance, rate):
        self.balance = balance
        self.rate = rate  

    def calculate_interest(self):
        return (self.balance * self.rate) / 100

account=SavingsAccount(50000,5)

print("Balance:", account.balance)
print("Interest Rate:", account.rate,"%")
print("Interest:", account.calculate_interest())
```

```python
# 10. Class Name: Notification
# Problem
# Create an abstract class named Notification.
# Explanation
# Notifications can be sent in different ways.
# Create abstract method:
# • send(message)
# Create subclasses:
# • EmailNotification
# • SMSNotification

from abc import ABC, abstractmethod
class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass

class EmailNotification(Notification):
    def send(self, message):
        print("Email Notification:",message)

class SMSNotification(Notification):
    def send(self, message):
        print("SMS Notification:", message)

email = EmailNotification()
sms = SMSNotification()

email.send("Welcome to our service")
sms.send("Your OTP is 123456")
```

```python
# 11. Class Name: FileHandler
# Problem
# Create an abstract class named FileHandler.
# Explanation
# Files can be read and written differently.
# Create abstract methods:
# • read()
# • write(data)
# Create subclasses like:
# • TextFileHandler
# • CSVFileHandler

from abc import ABC,abstractmethod
class FileHandler(ABC):
    @abstractmethod
    def read(self):
        pass

    @abstractmethod
    def write(self,data):
        pass
        
class TextFileHandler(FileHandler):
    def read(self):
        return f"Reading plain text data"

    def write(self,data):
        print(f"Writing plain text: {data}")

class CSVFileHandler(FileHandler):
    def read(self):
        return f"Reading CSV rows"

    def write(self,data):
        print(f"Writing CSV data: {data}")
        
text=TextFileHandler()
text.read()
csv=CSVFileHandler()
csv.write("deleted file")
```

```python
# 13. Class Name: Calculator
# Problem
# Create an abstract class named Calculator.
# Explanation
# Different calculators perform different operations.
# Create abstract method:
# • compute(a, b)
# Create subclasses:
# • AdditionCalculator
# • MultiplicationCalculator
from abc import ABC,abstractmethod
class Calculator(ABC):
    @abstractmethod
    def compute(self,a,b):
        pass

class AdditionCalculator(Calculator):
    def compute(self,a,b):
        return a + b

class MultiplicationCalculator(Calculator):
    def compute(self,a,b):
        return a * b

add=AdditionCalculator()
mult=MultiplicationCalculator()
print(add.compute(20,30))
print(mult.compute(5,2))
```

```python
# 14. Class Name: Database
# Problem
# Create an abstract class named Database.
# Explanation
# Different databases connect and execute queries differently.
# Create abstract methods:
# • connect()
# • query(sql)
# Create subclasses:
# • MySQLDatabase
# • MongoDatabase
from abc import ABC,abstractmethod
class Database(ABC):
    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def query(self,sql):
        pass
class MySQLDatabase(Database):
    
    def connect(self):
        print("MySQL Database")
        
    def query(self,sql):
        print(f"MySQL query: {sql}")

class MongoDatabase(Database):
    
    def connect(self):
        print("Connecting to MongoDB")
        
    def query(self,sql):
        print(f"MongoDB(from SQL/command): {sql}")
    
mysql=MySQLDatabase()
mysql.connect()
mysql.query("solve the question based on the query")
```

```python
# 15. Class Name: Shape3D
# Problem
# Create an abstract class named Shape3D.
# Explanation
# Three-dimensional shapes have different formulas.
# Create abstract methods:
# • volume()
# • surface_area()
# Create subclasses:
# • Cube
# • Sphere


from abc import ABC, abstractmethod
class Shape3D(ABC):
    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass

class Cube(Shape3D):
    def __init__(self,side):
        self.side=side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6*(self.side ** 2)

class Sphere(Shape3D):
    def __init__(self,radius):
        self.radius=radius

    def volume(self):
        return (4 / 3) *3.14* (self.radius ** 3)

    def surface_area(self):
        return 4 *3.14* (self.radius ** 2)

cube=Cube(5)
print("Cube Volume:",cube.volume())
print("Cube Surface Area:",cube.surface_area())
print("-"*25)
sphere=Sphere(3)
print("Sphere Volume:",sphere.volume())
print("Sphere Surface Area:",sphere.surface_area())
```
