```python
# 11. Define class Counter with class attribute count = 0. Add method to increment
# count.
class Counter:
    # Class attribute
    count = 0

    # Method to increment count
    def increment(self):
        Counter.count += 1
        print("Current Count:", Counter.count)


# Creating objects
c1 = Counter()
c2 = Counter()

# Incrementing count
c1.increment()   # Count = 1
c2.increment()   # Count = 2
c1.increment()   # Count = 3

# Display final count
print("Final Count:", Counter.count)
    
```

```python
# 15. Show difference between class attribute and instance attribute with example of
class Player:
    game = "Cricket"

    def __init__(self, name, score):
        
        self.name = name
        self.score = score

    def display(self):
        print("Player Name :", self.name)
        print("Score       :", self.score)
        print("Game        :", Player.game)
        print("="*25)


player1 = Player("Virat", 95)
player2 = Player("Rohit", 120)


player1.display()
player2.display()
```

```python
class Player:
    # Class attribute (shared by all objects)
    game = "Cricket"

    def __init__(self, name, score):
        # Instance attributes (different for each object)
        self.name = name
        self.score = score

# Create objects
p1 = Player("Virat", 95)
p2 = Player("Rohit", 80)

# Access class attribute
print("Game:", Player.game)

# Access instance attributes
print("Player 1:", p1.name, p1.score)
print("Player 2:", p2.name, p2.score)

# Modify class attribute
Player.game = "Football"

print("\nAfter changing class attribute:")
print("Player 1 Game:", p1.game)
print("Player 2 Game:", p2.game)

# Modify instance attribute
p1.score = 100

print("\nAfter changing instance attribute:")
print("Player 1 Score:", p1.score)
print("Player 2 Score:", p2.score)
```

```python
# 24. In class BankAccount, add getter and setter for private __balance with validation.
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance    

    
    def get_balance(self):
        return self.__balance

   
    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
            print("Balance updated successfully.")
        else:
            print("Invalid balance! Balance cannot be negative.")


account = BankAccount(5000)

print("Current Balance:", account.get_balance())

account.set_balance(8000)
print("Updated Balance:", account.get_balance())

account.set_balance(-1000)
print("Final Balance:", account.get_balance())
```
