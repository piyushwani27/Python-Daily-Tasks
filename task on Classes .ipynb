```python
# 1. Create a class Student with instance attributes name and age. Create two objects and print their details.
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

student1=Student("Piyush Pramod Wani",23)
student2=Student("Sahil Ramlal Vaidya",22)

print("Student 1: Details")
print("Name:",student1.name)
print("Age:",student1.age)
print("-"*25)
print("Student 2:Details")
print("Name:",student2.name)
print("Age:",student2.age)
```

```python
# 2. Define class Car with attributes brand, model, and year. Create an object and
# display all attributes.
class Car:
    def __init__(self,brand,model,year):
        self.brand=brand
        self.model=model
        self.year=year

details=Car("Volvo","e2x",2014)

print("Car Details:")
print("Car Brand:",details.brand)
print("Car Model:",details.model)
print("Car Year:",details.year)
```

```python
# 3. Create class Book with attributes title, author, price. Add a method display_info().
class Book:
    def __init__(self,title,author,price):
        self.title=title
        self.author=author
        self.price=price

    def display_info(self):
        print(f"Title Of the Book:  {self.title}")
        print(f"Author Of The Book: {self.author}")
        print(f"Price of the book:  {self.price}")
details=Book("Python Programming", "Guido van Rossum", 599)

details.display_info()
```

```python
# 4. Create class Employee with attributes emp_id, name, salary. Create 3 objects.
class Employee:
    def __init__(self,emp_id,name,salary):
        self.emp_id=emp_id
        self.name=name
        self.salary=salary
        
    def display_info(self):
        print(f"Employee Id:  {self.emp_id}")
        print(f"Name of the Employee: {self.name}")
        print(f"Employee Salary:  {self.salary}")   

Emp1=Employee(101,"Sarthak R.Khawaskar",25000)
Emp2=Employee(102,"Vijayraj S.Sangadikar",45000)
Emp3=Employee(103,"Sandesh K.Meshram",50000)

Emp1.display_info()
print("-"*50)
Emp2.display_info()
print("-"*50)
Emp3.display_info()
```

```python
# 6. Create class Rectangle with attributes length and width. Add method to calculate
# area.
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width

    def calculate_rectanangle(self):
        Area=self.length * self.width
        
        print(f"Length of the Rectangle:{self.length}")
        print(f"Width of the Rectangle:{self.width}")
        print(f"Area:{Area}")

rect1=Rectangle(25,10)

rect1.calculate_rectanangle()                
```
