#### 4. Count the number of lines in a text file.

```python
try:
    with open("task.txt", "r") as file:
        count = 0
        for line in file:
            count += 1
    print("Total number of lines:", count)
except FileNotFoundError:
    print("File not found.")
```

#### 11. Take student name and marks from user and store them in a file.

```python
try:
    with open("student.txt", "w") as file:
        n = int(input("Enter the number of students: "))

        for i in range(n):
            name = input("Enter student name: ")
            marks = int(input("Enter student marks: "))

            file.write(f"{name} {marks}\n")

    print("Student details stored successfully.")

except ValueError:
    print("Please enter valid marks.")
```

#### 13. Write a program to reverse the content of a file and save it in another file.

```python
try:
    with open("task.txt", "r") as source:
        data = source.read()

    reverse_data = data[::-1]

    with open("student.txt", "w") as destination:
        destination.write(reverse_data)

    print("Content reversed and saved successfully.")

except FileNotFoundError:
    print("File not found.")
```

#### 14. Count how many times a particular word appears in a file.

#### 15. Read a file and display the longest word in it.

```python
try:
    with open("task.txt", "r") as file:
        words = file.read().split()

    longest = max(words, key=len)

    print("Longest word:", longest)

except FileNotFoundError:
    print("File not found.")
```

#### 16. Write a program to merge two text files into a third file.

```python
try:
    with open("file1.txt", "r") as file1:
        data1 = file1.read()
    with open("file2.txt", "r") as file2:
        data2 = file2.read()
    with open("merged.txt", "w") as file3:
        file3.write(data1)
        file3.write("\n")
        file3.write(data2)
    print("Files merged successfully.")
except FileNotFoundError:
    print("One or both files not found.")
```

#### 17. Store a list of numbers in a file, then read and calculate their sum.

```python
try:
    numbers = [10, 20, 30, 40, 50]

    with open("numbers.txt", "w") as file:
        for num in numbers:
            file.write(str(num) + "\n")

    total = 0
    with open("numbers.txt", "r") as file:
        for line in file:
            total += int(line.strip())

    print("Numbers stored in file:", numbers)
    print("Sum of numbers:", total)

except FileNotFoundError:
    print("File not found.")

except ValueError:
    print("Invalid data found in the file.")
```

#### 18. Create a file and write 10 lines. Then read only the first 5 lines.

```python
try:
    with open("sample.txt", "w") as file:
        for i in range(1, 11):
            file.write(f"This is line {i}\n")

    with open("sample.txt", "r") as file:
        print("First 5 lines of the file:")
        for i in range(5):
            print(file.readline().strip())

except FileNotFoundError:
    print("File not found.")

except Exception as e:
    print("Error:", e)
```

#### 19. Write a program to replace a word in a file with another word.

```python
try:
    with open("sample.txt", "w") as file:
        file.write("Python is easy to learn.\n")
        file.write("Python is a popular programming language.\n")

    with open("sample.txt", "r") as file:
        content = file.read()

    old_word = "Python"
    new_word = "Java"

    content = content.replace(old_word, new_word)

    with open("sample.txt", "w") as file:
        file.write(content)

    with open("sample.txt", "r") as file:
        print("Updated File Content:")
        print(file.read())

except FileNotFoundError:
    print("File not found.")

except Exception as e:
    print("Error:", e)
```
