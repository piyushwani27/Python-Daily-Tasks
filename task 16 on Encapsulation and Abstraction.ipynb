```python
# 1. Class Name: BankAccount
# Problem
# Create a class named BankAccount.
# Explanation
# A bank account stores money. The balance should not be directly accessible from outside
# the class, so create a private attribute named __balance.
# Implement the following methods:
# • deposit(amount)
# o Deposit money into the account.
# o Amount must be greater than 0.
# • withdraw(amount)
# o Withdraw money.
# o Withdrawal should happen only if:
# ▪ amount > 0
# ▪ amount <= available balance

class BankAccount:
    def __init__(self):
        self.__balance = 0   # Private attribute

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Deposit amount must be greater than 0.")

    # Withdraw method
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be greater than 0.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    # Display current balance
    def display_balance(self):
        print(f"Current Balance: ₹{self.__balance}")


# Creating object
account = BankAccount()

# Deposit money
account.deposit(5000)
account.display_balance()

# Withdraw money
account.withdraw(2000)
account.display_balance()

# Invalid cases
account.deposit(-100)
account.withdraw(6000)
account.withdraw(-500)
```

```python
# 7. Class Name: Car
# Problem
# Create class Car.
# Explanation
# Store privately
# • __model
# • __year
# Provide
# • get_model()
# • set_model()
# • get_year()
# • set_year()
# Validation:
# Year cannot be greater than current year.

from datetime import datetime

class Car:
    def __init__(self, model, year):
        self.__model = model      # Private attribute
        self.__year = year        # Private attribute

    # Getter for model
    def get_model(self):
        return self.__model

    # Setter for model
    def set_model(self, model):
        self.__model = model
        print("Model updated successfully.")

    # Getter for year
    def get_year(self):
        return self.__year

    # Setter for year
    def set_year(self, year):
        current_year = datetime.now().year
        if year <= current_year:
            self.__year = year
            print("Year updated successfully.")
        else:
            print("Invalid year! Year cannot be greater than the current year.")


# Creating object
car = Car("Toyota", 2022)

# Display details
print("Model:", car.get_model())
print("Year:", car.get_year())

# Update model
car.set_model("Honda")

# Update year (Valid)
car.set_year(2025)

# Update year (Invalid)
car.set_year(2030)

# Display updated details
print("Updated Model:", car.get_model())
print("Updated Year:", car.get_year())
```

```python
# 8. Class Name: User
# Problem
# Create class User.
# Explanation
# Password must remain private.
# Private attribute
# __password
# Method
# change_password(new_password)
# Validation
# Password length ≥ 8
class User:
    def __init__(self, password):
        if len(password) >= 8:
            self.__password = password
        else:
            self.__password = ""
            print("Password must be at least 8 characters long.")

    def change_password(self, new_password):
        if len(new_password) >= 8:
            self.__password = new_password
            print("Password changed successfully.")
        else:
            print("Password must be at least 8 characters long.")

    def display_password(self):
        print("Password:", self.__password)


# Example
user1 = User("mypassword")
user1.display_password()

user1.change_password("newpass123")
user1.display_password()

user1.change_password("abc")
```

```python
# 9. Class Name: Rectangle
# Problem
# Create class Rectangle.
# Explanation
# Private attributes
# • __length
# • __width
# Methods
# • set_length()
# • get_length()
# • set_width()
# • get_width()
# • area()
# Validation
# Length and width must be positive.

class Rectangle:
    def __init__(self, length, width):
        if length > 0:
            self.__length = length
        else:
            self.__length = 1
            print("Length must be positive.")

        if width > 0:
            self.__width = width
        else:
            self.__width = 1
            print("Width must be positive.")


    def set_length(self, length):
        if length > 0:
            self.__length = length
        else:
            print("Length must be positive.")

    def get_length(self):
        return self.__length

    def set_width(self, width):
        if width > 0:
            self.__width = width
        else:
            print("Width must be positive.")

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width


rect = Rectangle(10, 5)

print("Length:", rect.get_length())
print("Width:", rect.get_width())
print("Area:", rect.area())

rect.set_length(15)
rect.set_width(8)

print("\nAfter updating:")
print("Length:", rect.get_length())
print("Width:", rect.get_width())
print("Area:", rect.area())

rect.set_length(-4)
rect.set_width(0)
```

```python
# 16. Class Name: Shape
# Problem
# Create an abstract class Shape.
# Explanation
# Declare abstract methods
# • calculate_area()

# Sumit Sir Notes
# • calculate_perimeter()
# Create subclasses like
# • Circle
# • Rectangle
# Implement both methods.
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return math.pi * self.radius ** 2

    def calculate_perimeter(self):
        return 2 * math.pi * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)

circle = Circle(7)
print("Circle Area:", circle.calculate_area())
print("Circle Perimeter:", circle.calculate_perimeter())

rectangle = Rectangle(10, 5)
print("Rectangle Area:", rectangle.calculate_area())
print("Rectangle Perimeter:", rectangle.calculate_perimeter())
```

```python
# 21. Class Name: BankAccount
# Problem
# Create abstract class BankAccount.
# Explanation
# Abstract method

# Sumit Sir Notes
# calculate_interest()
# Implement in
# • SavingsAccount
# • CurrentAccount
# Each account type should use a different interest calculation.
from abc import ABC, abstractmethod

class BankAccount(ABC):

    def __init__(self, balance):
        self.balance = balance

    @abstractmethod
    def calculate_interest(self):
        pass

class SavingsAccount(BankAccount):

    def calculate_interest(self):
        interest = self.balance * 0.05
        return interest

class CurrentAccount(BankAccount):

    def calculate_interest(self):

        interest = self.balance * 0.02
        return interest

savings = SavingsAccount(10000)
current = CurrentAccount(10000)

print("Savings Account Interest:", savings.calculate_interest())
print("Current Account Interest:", current.calculate_interest())
```

```python
# 23. Class Name: Calculator
# Problem
# Create abstract class Calculator.
# Explanation
# Abstract method
# compute(a, b)
# Implement in
# • AddCalculator
# • MultiplyCalculator
# Each subclass performs a different operation.
from abc import ABC, abstractmethod

class Calculator(ABC):

    @abstractmethod
    def compute(self, a, b):
        pass

class AddCalculator(Calculator):

    def compute(self, a, b):
        return a + b

class MultiplyCalculator(Calculator):

    def compute(self, a, b):
        return a * b

add = AddCalculator()
multiply = MultiplyCalculator()

print("Addition:", add.compute(10, 5))
print("Multiplication:", multiply.compute(10, 5))
```

```python
from abc import ABC, abstractmethod
import math


class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass

class Cube(Shape3D):
    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * (self.side ** 2)

class Sphere(Shape3D):
    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4 / 3) * math.pi * (self.radius ** 3)

    def surface_area(self):
        return 4 * math.pi * (self.radius ** 2)

class Cylinder(Shape3D):
    def __init__(self, radius, height):
        self.radius =radius
        self.height =height

    def volume(self):
        return math.pi * (self.radius ** 2) * self.height

    def surface_area(self):
        return 2 * math.pi * self.radius * (self.radius + self.height)

cube = Cube(4)
sphere = Sphere(3)
cylinder = Cylinder(3, 5)

print("Cube Volume:", cube.volume())
print("Cube Surface Area:", cube.surface_area())

print()

print("Sphere Volume:", sphere.volume())
print("Sphere Surface Area:", sphere.surface_area())

print()

print("Cylinder Volume:", cylinder.volume())
print("Cylinder Surface Area:", cylinder.surface_area())
```
