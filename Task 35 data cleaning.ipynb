#### 16.Clean the Debt to Income column where some values are written like 35%
#### and 0.4x. Convert the valid values into a consistent numeric format.

```python
def clean_dti(x):
    x = str(x).strip()

    if x.endswith("%"):
        return float(x[:-1]) / 100

    if x.endswith("x"):
        return float(x[:-1])

    return pd.to_numeric(x, errors="coerce")

df["debt_to_income"].apply(clean_dti)
```
