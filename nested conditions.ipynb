### 1. Take a number as input. If the number is positive, check whether it is even or odd and print the result.

```python
num=int(input("Enter the Number:"))
if num >0:
    if num  %2==0:
        print("Even No:",num)
    else:
        print("Odd No:",num)
else:
    print("Not A Positive No:",num)
```

### 11.Take a year as input. Check if it is a leap year using nested conditions.

```python
year=int(input("Enter the Year:"))
if year % 4==0:
    if year % 100==0:
        if year % 400==0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
            
```
