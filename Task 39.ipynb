```python
df["Annual_Salary"]=df["Salary"]*12
print(df)
```

```python
def experience_level(exp):
    if exp==0:
        return "Fresher"
    elif exp <= 2:
        return "Junior"
    else :
        return "Senior"
df["Experience_level"]=df["Experience"].apply(experience_level)
print(df)
```

```python
df_sorted=df.sort_values(
    by="Salary",
    ascending=False
)
print(df_sorted)
```
