```python
# Take three numbers from user.
# Print the largest number using 'and ' 'or'operator.

num1= int(input("Enter the number 01:"))
num2= int(input("Enter the number 02:"))
num3= int(input("Enter the number 03:"))

if num1>num2 and num2>num3:
    print("Number 01 is Greater:",num1)
elif num2>num3 and num3>num1 or num2==num1:
    print("Number 02 is Greater:",num2)
else:
    print("Number 03 is greater:",num3)
```

## Question 07:

```python
# Take a word from user.
# Check if the word is a palindrome or not.
# (Example:"madam","radar")

word =input("Enter the word:")
if word == word[::-1]:
    print("The word is a Palindrome.")
else:
    print("The word is not a palindrome.")
```

## 01.Logical Operations:

```python
# Take three subject marks from the user (English,Math,Science)
# Check if the student has passed:
# All marks should be >= 40(use 'and'operator)
# If any one subject is <40,print'Fail'

English = float(input("Enter the marks of English:"))
Math = float(input("Enter the marks of Math:"))
Science =float(input("Enter the marks of Science:"))

if English >=40 and Math >=40 and Science >=40:
    print("The student is passed.")
else:
    print("The student is Fail.")
```

## 02. Logical + User Input:

```python
# Take numbers from user.
# Check Whetehr the number 
# Positive AND Even 
# OR
# Negative OR Odd.
# print Appropriate Message.

number =float(input("Enter the number:"))

if number > 0 and number % 2==0:
    print("The number is Positive and Even.")
elif number < 0 or number % 2!=0:
    print("The number is Negative and odd.")
else:
    print("NUMBER NOT DEFINED!.")
```
