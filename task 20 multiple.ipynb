```python
# 1. Class Name: Person → Student
# Problem
# Create a parent class named Person and a child class named Student.
# Explanation
# The Person class should store:
# • name
# • age
# The Student class should inherit from Person and add:
# • marks
# Create a method in Student to display the student's name, age, and marks.

# Parent class
class Person:
    def _init_(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def _init_(self, name, age, marks):
        super()._init_(name, age)
        self.marks = marks

    def display(self):
        print("Name :", self.name)
        print("Age  :", self.age)
        print("Marks:", self.marks)
student1 = Student("Himanshu", 22, 95)

student1.display()
```

```python
# 2. Class Name: Vehicle → Car
# Problem
# Create a parent class named Vehicle.
# Explanation
# The Vehicle class should contain a method:
# • start()
# Create a child class Car that:
# • Adds a brand attribute.
# • Overrides the start() method to display a car-specific starting message.

# Parent class
class Vehicle:
    def start(self):
        print("Vehicle is starting...")

# Child class
class Car(Vehicle):
    def _init_(self, brand):
        self.brand = brand

    # Overriding the start() method
    def start(self):
        print(f"{self.brand} car is starting with a key...")

# Create object
car1 = Car("Toyota")

# Call method
car1.start("car is Driving")
```

```python
# 3. Class Name: Animal → Dog
# Problem
# Create a parent class named Animal.
# Explanation

# Sumit Sir Notes
# The Animal class should contain:
# • eat()
# Create a child class Dog that:
# • Inherits from Animal.
# • Adds a method:
# o bark()
# Demonstrate both methods.

# Parent class
class Animal:
    def eat(self):
        print("Animal is eating.")

# Child class
class Dog(Animal):
    def bark(self):
        print("Dog is barking.")

# Create object
dog1 = Dog()

# Demonstrate both methods
dog1.eat()
dog1.bark()
```

```python
# 4. Class Name: Employee → Manager
# Problem
# Create a parent class named Employee.
# Explanation
# The Employee class should store:
# • emp_id
# • salary
# Create child class Manager.
# Add:
# • department
# Display all employee information.

# Parent class
class Employee:
    def _init_(self, emp_id, salary):
        self.emp_id = emp_id
        self.salary = salary

# Child class
class Manager(Employee):
    def _init_(self, emp_id, salary, department):
        super()._init_(emp_id, salary)
        self.department = department

    def display(self):
        print("Employee ID :", self.emp_id)
        print("Salary      :", self.salary)
        print("Department  :", self.department)

# Create object
manager1 = Manager(101, 50000, "IT")

# Display employee information
manager1.display()
```

```python
# 5. Class Name: Shape → Circle
# Problem
# Create a parent class named Shape.
# Explanation
# Create a method:
# • area()
# Create child class:
# • Circle

# Sumit Sir Notes
# Store:
# • radius
# Override area() to calculate the area of a circle.

# Parent class
class Shape:
    def area(self):
        print("Area of shape")

# Child class
class Circle(Shape):
    def _init_(self, radius):
        self.radius = radius

    # Overriding the area() method
    def area(self):
        area = 3.14 * self.radius * self.radius
        print("Area of Circle:", area)

# Create object
circle1 = Circle(5)

# Call method
circle1.area()
```
