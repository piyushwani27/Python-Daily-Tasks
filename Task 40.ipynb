```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")

# Scatter plot showing bill vs tip, colored by day
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()
```

```python
fmri = sns.load_dataset("fmri")

# Line plot tracking signal over timepoints for different regions
sns.lineplot(data=fmri, x="timepoint", y="signal", hue="region")
plt.title("Line Plot of FMRI Signal Over Time")
plt.show()
```

```python
# Histogram and density curve of total bill amounts
sns.histplot(data=tips, x="total_bill", kde=True)
plt.title("Data Distribution of Total Bill")
plt.show()
```

```python
# Load the classic 'iris' flower dataset
iris = sns.load_dataset("iris")

# Matrix grid of all numerical variables grouped by plant species
sns.pairplot(data=iris, hue="species")
plt.show()
```

```python
# Box plot showing the spread of tips by day of the week
sns.boxplot(data=tips, x="day", y="tip")
plt.title("Box Plot of Tips by Day")
plt.show()
```

```python
# Bar plot comparing average total bill by day
sns.barplot(data=tips, x="day", y="total_bill")
plt.title("Average Total Bill by Day")
plt.show()
```

```python
# Compute correlation matrix for numeric columns
numeric_iris = iris.drop(columns="species")
corr_matrix = numeric_iris.corr()

# Plot the heatmap with annotations
sns.heatmap(data=corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix Heatmap")
plt.show()
```
