```python
# 9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a
# nested element and checking the original.
import copy

def detect_copy(copy_func):
    # Original nested list
    original = [[1, 2], [3, 4]]

    # Create copy using the given function
    copied = copy_func(original)

    # Modify a nested element in the copied list
    copied[0][0] = 99

    # Check whether original changed
    if original[0][0] == 99:
        print("Shallow Copy Detected")
    else:
        print("Deep Copy Detected")

    print("Original:", original)
    print("Copied  :", copied)


# Test with shallow copy
print("Testing Shallow Copy:")
detect_copy(copy.copy)

print("\nTesting Deep Copy:")
detect_copy(copy.deepcopy)
```

```python
# 10. (Hard) Create a class Student with attributes name and marks (list). Demonstrate what
# happens when you do shallow copy vs deep copy of a Student object.

import copy

# Student class
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks                       # Original student object
student1 = Student("Piyush", [85, 90, 95])      # Create copies
shallow_student = copy.copy(student1)
deep_student = copy.deepcopy(student1)          # Modify the marks list in the shallow copy

shallow_student.marks.append(100)
print("After modifying shallow copy:")
print("Original Student Marks :", student1.marks)
print("Shallow Copy Marks     :", shallow_student.marks)
print("Deep Copy Marks        :", deep_student.marks)   # Modify the marks list in the deep copy

deep_student.marks.append(110)
print("\nAfter modifying deep copy:")
print("Original Student Marks :", student1.marks)
print("Shallow Copy Marks     :", shallow_student.marks)
print("Deep Copy Marks        :", deep_student.marks)
```

```python
# 3. Mobile
# Create a Mobile class.
# Attributes
# • brand
# • model
# • price
# Create a method that applies a 10% discount and updates price.

class Mobile():
    def __init__(self,brand,model,price):
        self.brand=brand
        self.model=model
        self.price=price
        
    def apply_discount(self):
        self.price = self.price - (self.price * 10 / 100)
        print("Price after 10% discount: ₹", self.price)

mobile1 = Mobile("Samsung", "Galaxy S24", 80000)
print("Brand :", mobile1.brand) 
print("Model :", mobile1.model) 
print("Original Price: ₹", mobile1.price)
mobile1.apply_discount()
```

```python
# Create a Laptop class

class Laptop:
    def __init__(self, brand, ram, storage):
        self.brand = brand
        self.ram = ram          # RAM in GB
        self.storage = storage  # Storage in GB

    # Method to upgrade RAM by 8 GB
    def upgrade_ram(self):
        self.ram += 8
        print("RAM upgraded successfully!")

    # Method to display laptop details
    def display(self):
        print("Brand:", self.brand)
        print("RAM:", self.ram, "GB")
        print("Storage:", self.storage, "GB")


# Create an object
laptop1 = Laptop("Dell", 8, 512)

print("Before RAM Upgrade:")
laptop1.display()

# Upgrade RAM
laptop1.upgrade_ram()

print("\nAfter RAM Upgrade:")
laptop1.display()
```

```python
# 5. Bank Account
# Attributes
# • account_holder
# • balance
# Methods
# • deposit()
# • withdraw()
# Withdraw should happen only if balance is sufficient.

class Bank_Account():
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
    # Method 01:Deposite()
    def deposit(self,amount):
        self.balance=+amount
        print("₹", amount, "deposited successfully")

    # method 02():withdrawl money
    def withdrawl(self,amount):
        if amount <=self.balance:
            self.balance-=amount
            print("₹", amount, "withdrawn successfully.")
        else:
            print("Insufficient balance!")
    def display(self):
        print("Account Holder:",self.account_holder)
        print("Balance:",self.balance)
account1=Bank_Account("Piyush Wani",10000)
print("Initial account Details:")
account1.display()
account1.deposit(5000)

# Withdraw money
account1.withdrawl(3000)

# Try to withdraw more than the balance
account1.withdrawl(15000)

print("\nFinal Account Details:")
account1.display()
```

```python
# Create a BankAccount class

class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    # Method to deposit money
    def deposit(self, amount):
        self.balance += amount
        print("₹", amount, "deposited successfully.")

    # Method to withdraw money
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("₹", amount, "withdrawn successfully.")
        else:
            print("Insufficient balance!")

    # Method to display account details
    def display(self):
        print("Account Holder:", self.account_holder)
        print("Balance: ₹", self.balance)


# Create an object
account1 = BankAccount("Piyush", 10000)

print("Initial Account Details:")
account1.display()

# Deposit money
account1.deposit(5000)

# Withdraw money
account1.withdraw(3000)

# Try to withdraw more than the balance
account1.withdraw(15000)

print("\nFinal Account Details:")
account1.display()
```

```python
# Create a Car class

class Car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    # Method to increase speed
    def increase_speed(self, value):
        if self.speed + value <= 200:
            self.speed += value
            print("Speed increased successfully.")
        else:
            self.speed = 200
            print("Maximum speed limit reached (200 km/h).")

    # Method to display car details
    def display(self):
        print("Brand:", self.brand)
        print("Speed:", self.speed, "km/h")


# Create an object
car1 = Car("Toyota", 180)

print("Before Increasing Speed:")
car1.display()

# Increase speed
car1.increase_speed(15)

print("\nAfter First Increase:")
car1.display()

# Try to exceed maximum speed
car1.increase_speed(20)

print("\nAfter Second Increase:")
car1.display()
```
