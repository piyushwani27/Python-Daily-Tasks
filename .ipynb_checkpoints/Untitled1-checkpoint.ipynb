```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load built-in dataset
tips = sns.load_dataset("tips")

# Create bar plot
sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

# Add title
plt.title("Average Total Bill by Day")

# Display graph
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [80, 120, 60, 90]

sns.barplot(x=products, y=sales)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.title("Average Bill by Day and Gender")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [80, 120, 60, 90]

sns.barplot(
    x=sales,
    y=products
)

plt.title("Product Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    errorbar=None
)

plt.title("Average Total Bill by Day")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    errorbar=None
)

plt.title("Average Bill: Smoker vs Non-Smoker")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")

plt.show()
```
