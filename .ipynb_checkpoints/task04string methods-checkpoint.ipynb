### 01. Take a string from user and replace all spaces with underscore _ using replace() method.


```python
string = input("Enter the string:")
print(string.replace(" ","_"))
```

### 02.Take any string and remove leading and trailing spaces using strip().

```python
name =("     Piyush Pramod Wani     ")
print(name.strip())
print(name.lstrip())
print(name.rstrip())
```
