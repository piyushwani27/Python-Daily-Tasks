## 6. Comparison Operator:

```python
# Take Two Numbers and check
# Equal ,Not Equal,Greater,Smaller,Greater or equal,smaller or equal

X = int(input("Enter th no X:"))
Y = int(input("Enter th no Y:"))

print(X == Y) #Equal
print(X != Y) #Not Equal
print(X > Y) #Greater
print(X < Y) #Smaller
print(X >= Y) #Greater or Equal
print(X <= Y) #Smaller or Equal
```

## 9.Simple Calculator :

```python
# Create a simple calculator that takes two numbers and one operator(+,-,*,/,//,%,**) 
# from the user and shows the result.

num1 = float(input("Enter the no Num1 :"))
num2 = float(input("Enter the no Num2:"))

operator = input("Enter operator(+,-,*,/,//,%,**):")
if operator=="+":
    print(num1 + num2)
elif operator =="-":
    print(num1 - num2)
elif operator =="*":
    print(num1 * num2)
elif operator =="/":
    if num2 !=0:
        print("Result:",num1/num2)
    else:
        print("Error:Division by zero!")
elif operator =="//":
    if num2!=0:
        print("Result:",num1//num2)
    else:
        print("Error:Division by Zero!")
elif operator =="%":
    if num2!=0:
        print("Result:",num1 % num2)
    else:
        print("Error :Division by Zero!")
elif operator =="**":
    if num2!=0:
        print("Result:",num1**num2)
    else:
        print("Error :Division by Zero!")
else:
    print("invalid operator")
```

## 12.Student Grade calculator:

```python
# Take marks of 5 Subjects.
#Calculate Percentage and Print grade.
#>=90=A+,>=80=A,>=70=B,>=60=C,else=Fail.
print("Enter the Marks of 5 Subjects:")
English =float(input("Enter the marks of English:"))
Marathi =float(input("Enter the marks of Marathi:"))
Hindi =float(input("Enter the marks of Hindi:"))
Science =float(input("Enter the marks of Science:"))
Maths =float(input("Enter the marks of Maths:"))

total_marks= English + Marathi + Hindi + Science + Maths
print(total_marks)
Percentage = (total_marks/500*100)

if Percentage >=90:
    Grade = "A+"
elif Percentage >=80:
    Grade = "A"
elif Percentage >=70:
    Grade ="B"
elif Percentage >=60:
    Grade ="C"
else :
    Grade ="Fail"

print(f"Total Marks:{total_marks}/500")
print(f"Percentage:{Percentage:.2f}%")
print(f"Grade:{Grade}")
```
