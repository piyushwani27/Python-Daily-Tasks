## String Slicing Questions:

### 01.Take String From user and print first 5 characters.

### 02. Take a string and print the last 4 character using negative slicing.

```python
string =input(" Enter the string:")
print("Last four Characters are:",string[-4:])
```

### 03.Take astring and print characters from 2 to 8 (excluding 8).

### 04.Take a String and print every 2nd character of the string.

```python
full_name =input("Enter your Full Name:")
print("Every 2nd Character of the String Are:",full_name[::2])
```

### 05.Take a full name and Print only the first name and last name separetely using slicing.

```python
full_name =("yash arun wakode")
print("First Name:",full_name[:4])
print("last Name:",full_name[-6:])
```

## Case Conversion Questions:

### 06.Take a String from User and Print it in:
* Uppercase
* Lowercase
* Titlecase
* Capitalize(Only First letter capital.)

```python
full_name =input("Enter the full name:")
print("Uppercase:",full_name.upper())
print("Lowercase:",full_name.lower())
print("Titlecase:",full_name.title())
print("Capitilize:",full_name.capitalize())
```

### 07. take a Sentence and convert it to title case. Then print the original and converted version.

```python
text =input("Enter the Text:")
print("Title case:",text.title())
print("Original Text:",text)
print("Converted version",text.title())
```

### 08.take a mixed case string and check if lower() and upper() methods hange the string permenantly or not.

```python
text =("sanKET SAhil SAmeer piyuSH")
# Mixed string

print("After Lower:",text.lower()) 
# After Changing the String in Lower Case.

print("After Upper:",text.upper())
# After Changing the String in Upper Case.

print("Original string:",text)

 # The loewr() and the upper() method do not change the strin permanently.
 # Because string imn python are immutable ,These methods return a nfew string .
 # Leaving the original string unchanged unlesss you assign the new variable.
```

### 13.Take a Password as input and check.
* it should be alphanumeric.
* it should be minimum 8 characters long print "valid Password" or "Invalid Password"

```python
password =input("Enter the Password:")
# Taking input from the User as a Password.

if password.isalnum() and len(password)>= 8:
    # if the length of the password is >=8 and the password contains alphanumeric the it will return the password.
    print("Valid Password.")
    # It will reutun the Password is Valid.
else:
    print("Invalid Password.")
    # It will return the Password is Invalid.if it doesn't contain the valid length and the alphabets and the digit only.
```

## Find() and index() Questions.

### 14. Take a string and a character from user. Find the first occurrence of that character using find(). 

```python
text = input("Enter the Text:")
char =input("Enter The Character:")

print(text.find(char))
```

### 15. Take a string and a word. Use find() to check if the word exists. If not found, print -1. 

```python
string =input("Enter the string:")
word =input("Enter the word:")

print(string.find(word))
```

## 16. Take a sentence and find the position of the word "Python" using index(). 

```python
sentence =("I love Python.")
word =("Python")

print(sentence.index(word))
```

 ## 17. Take a string. Use both find() and index() to search for a character that does not exist. Observe the difference.  

```python
string = input("Enter the String.")
char =input("Enter the character.")
print(string.find(char))
print(string.index(char))
```

### Mixed Questions 

## 18. Take your full name.
• Convert it to uppercase 
• Find the position of first space 
• Print the first name and last name separately. 

```python
full_name =("piyush pramod wani")
print("Uppercase:",full_name.upper())
print("Position Of First Space:",full_name.find(" "))
print("First Name:",full_name[:7])
print("Last Name:",full_name[-4:])
```

## 19. Take a mobile number as string. 
    Check if it is 10 digits and contains only numbers using isdigit() and len(). 

```python
number =input("Enter the mo no:")
if len(number)==10 and number.isdigit():
    print("IT is MO Number:",number)
else:
    print("IT is not a MO Number",number)
```
