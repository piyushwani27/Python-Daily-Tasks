```python
# 1. Person → Student, Teacher
# Question
# Create a parent class Person with attributes name and age. Create two child classes Student
# and Teacher that inherit from Person.
# Concept
# • Hierarchical Inheritance
# • Parent attributes inherited by multiple child classes.
# How to Solve
# 1. Create a parent class Person.
# 2. Add name and age using a constructor.
# 3. Create Student(Person).
# 4. Create Teacher(Person).
# 5. Create objects of both classes.
# 6. Show that both children can access name and age.
# Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def display(self):
        print("Student Details")
        print("Name:", self.name)
        print("Age:", self.age)

class Teacher(Person):
    def display(self):
        print("Teacher Details")
        print("Name:", self.name)
        print("Age:", self.age)

student = Student("Rahul", 20)
teacher = Teacher("Anita", 35)

student.display()
print()
teacher.display()
```

```python
# 2. Animal → Dog, Cat
# Question
# Define class Animal with method eat(). Create subclasses Dog and Cat.
# Concept
# • Method inheritance.
# How to Solve
# 1. Create parent class Animal.
# 2. Write method eat().
# 3. Create child classes Dog and Cat.
# 4. Create objects.
# 5. Call eat() using both objects.
# Parent Class
class Animal:
    def eat(self):
        print("Animal is eating.")

class Dog(Animal):
    pass

class Cat(Animal):
    pass

dog = Dog()
cat = Cat()

print("Dog:")
dog.eat()

print("\nCat:")
cat.eat()
```

```python
# 3. Vehicle → Car, Bike
# Question
# Create class Vehicle with method start(). Create subclasses Car and Bike.
# Concept
# • One parent method shared by multiple children.
# How to Solve
# 1. Create Vehicle.
# 2. Add start().
# 3. Inherit it in Car.
# 4. Inherit it in Bike.
# 5. Call start() from both child objects.
# Parent Class
class Vehicle:
    def start(self):
        print("Vehicle started.")

class Car(Vehicle):
    pass

class Bike(Vehicle):
    pass

car=Car()
bike=Bike()

print("Car:")
car.start()

print("\nBike:")
bike.start()
```

```python
# 4. Shape → Circle, Rectangle
# Question
# Define class Shape with method area(). Create subclasses Circle and Rectangle.
# Concept
# • Parent provides common method.
# • Children may later override it.
# How to Solve
# 1. Create Shape.
# 2. Add area().
# 3. Create Circle.
# 4. Create Rectangle.
# 5. Call area().
# Parent Class
class Shape:
    def area(self):
        print("Calculating area of a shape.")

class Circle(Shape):
    pass

class Rectangle(Shape):
    pass

circle=Circle()
rectangle=Rectangle()

print("Circle:")
circle.area()

print("\nRectangle:")
rectangle.area()
```

```python
# 5. Employee → Manager, Developer
# Question

# Sumit Sir Notes
# Create class Employee with attributes emp_id and salary. Create subclasses Manager and
# Developer.
# Concept
# • Attribute inheritance.
# How to Solve
# 1. Create parent constructor.
# 2. Store emp_id and salary.
# 3. Create two child classes.
# 4. Show inherited attributes.
# Parent Class
class Employee:
    def __init__(self, emp_id, salary):
        self.emp_id=emp_id
        self.salary=salary

class Manager(Employee):
    def display(self):
        print("Manager Details")
        print("Employee ID:", self.emp_id)
        print("Salary: ₹", self.salary)

class Developer(Employee):
    def display(self):
        print("Developer Details")
        print("Employee ID:", self.emp_id)
        print("Salary: ₹", self.salary)

manager=Manager(101, 75000)
developer=Developer(102, 60000)

manager.display()
print()
developer.display()
```

```python
# 6. BankAccount → SavingsAccount, CurrentAccount
# Question
# Define class BankAccount with method deposit().
# Concept
# • Common banking functionality.
# How to Solve
# 1. Create deposit().
# 2. Create two child classes.
# 3. Call deposit() from both.
# Parent Class
class BankAccount:
    def deposit(self):
        print("Amount deposited successfully.")

class SavingsAccount(BankAccount):
    pass

class CurrentAccount(BankAccount):
    pass

savings=SavingsAccount()
current=CurrentAccount()

print("Savings Account:")
savings.deposit()

print("\nCurrent Account:")
current.deposit()
```

```python
# 8. Sports → Cricket, Football
# Question
# Define class Sports with method play().
# Concept
# • Common method inherited.
# How to Solve
# 1. Parent class.
# 2. Common method.
# 3. Two subclasses.
# 4. Call inherited method.
# Parent Class
class Sports:
    def play(self):
        print("Playing a sport.")

class Cricket(Sports):
    pass

class Football(Sports):
    pass

c = Cricket()
f = Football()

print("Cricket:")
c.play()
print("\nFootball:")
f.play()
```

```python
# 9. Product → Electronics, Clothing
# Question
# Create class Product with attribute price.
# Concept
# • Common data shared by children.
# How to Solve
# 1. Create constructor.
# 2. Store price.
# 3. Create child classes.
# 4. Display price.
# Parent Class
class Product:
    def __init__(self, price):
        self.price=price

class Electronics(Product):
    def display(self):
        print("Electronics Product")
        print("Price: ₹", self.price)

class Clothing(Product):
    def display(self):
        print("Clothing Product")
        print("Price: ₹",self.price)

electronic_item=Electronics(45000)
clothing_item=Clothing(1999)

electronic_item.display()
print()
clothing_item.display()
```

```python
# 10. Book → Fiction, NonFiction
# Question
# Define class Book with attributes title and author.

# Sumit Sir Notes
# Concept
# • Constructor inheritance.
# How to Solve
# 1. Parent constructor.
# 2. Two child classes.
# 3. Access inherited attributes.
# Parent Class
class Book:
    def __init__(self, title, author):
        self.title=title
        self.author=author

# Child Class 1
class Fiction(Book):
    def display(self):
        print("Fiction Book")
        print("Title :", self.title)
        print("Author:", self.author)

class NonFiction(Book):
    def display(self):
        print("Non-Fiction Book")
        print("Title :", self.title)
        print("Author:", self.author)

fiction_book=Fiction("Harry Potter", "J. K. Rowling")
nonfiction_book=NonFiction("Wings of Fire", "A. P. J. Abdul Kalam")

fiction_book.display()
print()
nonfiction_book.display()
```
