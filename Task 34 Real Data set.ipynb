#### 18.Identify unrealistic age values using NumPy/Pandas and replace them with  NaN.  

```python
import numpy as np

df =df.copy()

df["age"]=pd.to_numeric(df["age"],errors="coerce")
df.loc[~df["age"].between(18, 100),"age"]=np.nan
print(df["age"])
```

#### 19.Clean the credit_score column by removing text such as "pts" and converting  the column to numeric. 

```python
df["credit_score"] = pd.to_numeric(
    df["credit_score"].astype(str).str.replace("pts", "", case=False).str.strip(),
    errors="coerce"
)

print(df["credit_score"])
```
