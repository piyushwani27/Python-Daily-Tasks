#### Challenge 1 – Complete Data Cleaning
#### Your manager gives you a messy dataset.
#### Perform all of the following:
#### • Load the dataset.
#### • Display dataset information.
#### • Find missing values.
#### • Remove duplicate rows.
#### • Fill missing values:
#### o Age → median
#### o Balance → mean
#### • Rename:
#### o Target → CustomerStatus
#### o AnnualIncome → Income
#### • Display the first 15 rows.
#### Methods expected:
#### read_csv()
#### info()
#### isnull()
#### drop_duplicates()
#### median()
#### mean()
#### fillna()
#### rename()
#### head()

```python
import pandas as pd

df = pd.read_csv("C:\\Users\\HP\\Downloads\\Task_dataset - Task_dataset.csv")

df.info()

print("Missing Values:")
print(df.isnull().sum())


df["Age"] = df["Age"].fillna(df["Age"].median())

df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

df = df.rename(columns={
    "Target": "CustomerStatus",
    "AnnualIncome": "Income"
})

df.head(15)
```
