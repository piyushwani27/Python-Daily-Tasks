##### 10. Create a dictionary of student marks. Use get(), keys(), values(), items(), update() and pop().

```python
student_marks = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 95
}
print("Original dictionary:", student_marks)

bob_marks = student_marks.get("Bob")
print("Bob's marks:", bob_marks)

all_students = student_marks.keys()
print("All keys:", all_students)

all_marks = student_marks.values()
print("All values:", all_marks)

all_items = student_marks.items()
print("All items:", all_items)

student_marks.update({"Charlie": 82, "Eva": 89})
print("After update():", student_marks)

david_mark = student_marks.pop("David")
print(f"Removed David's mark: {david_mark}")
print("After pop('David'):", student_marks)
```

##### 14. Create class BankAccount with private attribute __balance. Add deposit and withdraw methods with validation.

```python
class BankAccount:
    def __init__(self, initial_balance=0.0):
        # Initialize the private attribute
        self.__balance = float(initial_balance) if initial_balance >= 0 else 0.0

    # Method to deposit money with validation
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Successfully deposited ₹{amount:.2f}")
        else:
            print("Error: Deposit amount must be positive.")

    # Method to withdraw money with validation
    def withdraw(self, amount):
        if amount <= 0:
            print("Error: Withdrawal amount must be positive.")
        elif amount > self.__balance:
            print("Error: Insufficient funds.")
        else:
            self.__balance -= amount
            print(f"Successfully withdrew ₹{amount:.2f}")

    # Getter method to safely view the balance
    def get_balance(self):
        return self.__balance

# Demonstration
account = BankAccount(1000)  # Starting with ₹1000

# 1. Test Valid Deposit
account.deposit(500)

# 2. Test Invalid Deposit
account.deposit(-50)

# 3. Test Valid Withdrawal
account.withdraw(300)

# 4. Test Overdraft Withdrawal
account.withdraw(2000)

# Check Final Balance
print(f"Final Balance: ₹{account.get_balance():.2f}")
    
```
