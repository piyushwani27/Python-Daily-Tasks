###  18. Take 5 city names as input from user and store them in a list. Then print them in reverse order.

```python
city=[]

city.append(input("Enter the city Names 1:"))
city.append(input("Enter the city Names 2:"))
city.append(input("Enter the city Names 3:"))
city.append(input("Enter the city Names 4:"))
city.append(input("Enter the city Names 5:"))

print(city)
city.reverse()
print(city)
```

```python
cities =[]
print("Enter the 5 city names:")
for i in range(5):
    city= input(f"City {i+1}:")
    cities.append(city)
print("\nThe Cities in Reverse Order:")
for city in cities[::-1]:
    print(city)
```
