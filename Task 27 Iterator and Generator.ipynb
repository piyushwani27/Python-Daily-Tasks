#### Problem 6
#### Create a custom iterator that prints numbers from 1 to 10.

```python
class NumberIterator:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

numbers = NumberIterator()

for i in numbers:
    print(i)
```

# Problem 8
# Create a custom iterator that prints numbers in reverse order from
# # 10 to 1.

```python
class ReverseIterator:
    def __init__(self):
        self.num = 10

    def __iter__(self):
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration

reverse_numbers = ReverseIterator()

for i in reverse_numbers:
    print(i)
```

# Problem 9
# Create a custom iterator that prints the characters of the word
# "GENERATOR" one by one.

```python
class WordIterator:
    def __init__(self, word):
        self.word = word
        self.i = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.i < len(self.word):
            ch = self.word[self.i]
            self.i += 1
            return ch
        raise StopIteration


obj = WordIterator("GENERATOR")

for letter in obj:
    print(letter)
```

# Problem 10
# Create a custom iterator that prints the multiplication table of 5.
# Expected Output:

# 5
# 10
# 15
# 20
# 25
# 30
# 35
# 40
# 45
# 50

```python
class MultipleIterator:
    def __init__(self):
        self.i = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.i <= 10:
            result = 5 * self.i
            self.i += 1
            return result
        else:
            raise StopIteration

multiple = MultipleIterator()

for i in multiple:
    print(i)
```

# Level 4: Generator with Lists
# Problem 16
# Given the following list:
# numbers = [10, 15, 20, 25, 30, 35]
# Create a generator that yields only the even numbers.

```python
numbers = [10, 15, 20, 25, 30, 35]

def even_numbers(nums):
    for num in nums:
        if num % 2 == 0:
            yield num

gen = even_numbers(numbers)

for i in gen:
    print(i)
```

# Problem 17
# Create a generator that yields only the odd numbers from the list
# below.
# numbers = [1, 4, 7, 10, 13, 16, 19]

```python
numbers = [1, 4, 7, 10, 13, 16, 19]

def Odd_numbers(nums):
    for num in nums:
        if num % 2 != 0:
            yield num
            
gen = Odd_numbers(numbers)
for i in gen:
    print(i)
```

# Problem 19
# Given a list of numbers, create a generator that yields only numbers
# greater than 50.
# numbers = [20, 55, 40, 90, 60, 15, 75]

```python
numbers = [20, 55, 40, 90, 60, 15, 75]

def greater_than_50(nums):
    for num in nums:
        if num > 50:
            yield num

gen = greater_than_50(numbers)

for i in gen:
    print(i)
```

# Problem 20
# Create a generator that yields the length of each word in the
# following list.
# words = ["Python", "Java", "AI", "Generator", "Iterator"]

```python
words = ["Python", "Java", "AI", "Generator", "Iterator"]
def word_lengths(words):
    for word in words:
        yield len(word)

gen = word_lengths(words)

for length in gen:
    print(length)
```
