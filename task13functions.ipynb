```python
from datetime import datetime

def show_date():
    """Prints the current date and time."""
    now = datetime.now()
    # Formats to a readable string (e.g., YYYY-MM-DD HH:MM:SS)
    formatted_now = now.strftime("%Y-%m-%d %H:%M:%S")
    print("Current Date and Time:", formatted_now)

# Call the function to display the output
show_date()
```

```python
# 16. Create a function calculate(a, b, operation) and show how keyword args improve readability.
def calculate(a, b, operation):
    if operation == "add":
       return a + b
    elif operation == "subtract":
       return a - b
    elif operation == "multiply":
       return a * b
    elif operation == "divide":
       return a / b if b != 0 else "Cannot divide by zero"
    else:
       return "Unknown operation"
print(calculate(a=50,b=30,operation="add"))
```

```python
# 23. Write a function multiply(*args) that multiplies all numbers.
def multiply_all(*args):
    result = 1
    for num in args:
        result *= num
    return result

# Function calls
print(multiply_all(2, 3, 4))      # Output: 24
print(multiply_all(5, 10))        # Output: 50
print(multiply_all(1, 2, 3, 4, 5))# Output: 120
```

```python
# 28. Write a function calculate_total(**kwargs) where keys are item names and values are
# prices.
def calculate_total(**kwargs):
    total = 0

    for item, price in kwargs.items():
        print(item, ":", price)
        total += price

    return total

# Function call
total = calculate_total(
    Apple=50,
    Banana=30,
    Milk=60,
    Bread=40
)

print("Total Price =", total)
```
